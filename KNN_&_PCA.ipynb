# 1. What is K-Nearest Neighbors (KNN) and how does it work in both classification and regression problems?

- K-Nearest Neighbors (KNN)

K-Nearest Neighbors (KNN) is a simple, supervised machine learning algorithm used for both classification and regression. It makes predictions by finding the K training data points closest to a new data point and using their known outcomes to make the prediction.

KNN is called a non-parametric, instance-based algorithm because it does not build a traditional mathematical model during training; it essentially stores the training examples and makes decisions when a prediction is requested.

How KNN works

Suppose we choose K = 3.

1. Choose the value of K.

2. Calculate the distance between the new data point and the training points.

3. Identify the K nearest points.

4. Use those neighbors to make a prediction.

5. Commonly, Euclidean distance is used, although other distance metrics can also be used.

A common Euclidean distance formula is:

$$ d=\sqrt{(x_1-x_2)^2+(y_1-y_2)^2} $$
k = 5: Class A wins 3–2; the point is Class A.

1. KNN for Classification

In classification, the target variable consists of categories or classes, such as:

Spam / Not Spam

Pass / Fail

Cat / Dog

Loan Default / No Default

KNN assigns the new observation to the class that receives the majority vote among its K nearest neighbors.

Example:

If K = 5 and the nearest neighbors are:

3 → Class A

2 → Class B

Then the new observation is classified as Class A.

So:

Prediction = majority class of the K nearest neighbors

KNN can also give closer neighbors greater influence by using distance weighting rather than giving every neighbor equal weight.

2. KNN for Regression

In regression, the target variable is continuous, such as:

House price

Salary

Temperature

Sales revenue

Instead of voting for a class, KNN calculates the average target value of the K nearest neighbors.

Example:

Suppose K = 3 and the three nearest houses have prices:

₹40 lakh

₹45 lakh

₹50 lakh

The predicted price is:

$$ \frac{40+45+50}{3}=45\text{ lakh} $$

Therefore, the predicted house price is ₹45 lakh.

Distance-weighted regression can give more influence to neighbors that are closer to the new observation.

| Feature      | KNN Classification     | KNN Regression        |
| ------------ | ---------------------- | --------------------- |
| Target       | Categorical            | Continuous            |
| Prediction   | Majority vote          | Average of neighbors  |
| Example      | Spam/Not Spam          | House price           |
| Scikit-learn | `KNeighborsClassifier` | `KNeighborsRegressor` |
| Output       | Class label            | Numerical value       |

Importance of choosing K

The value of K has a major effect on the model:

Small K → sensitive to individual data points and noise; can lead to overfitting.

Large K → smoother predictions but may overlook local patterns and lead to underfitting.

In practice, K is usually selected by testing different values, often using cross-validation.


# 2. What is the Curse of Dimensionality and how does it affect KNN performance?

- Curse of Dimensionality

The Curse of Dimensionality refers to the problems that occur when the number of features (dimensions) in a dataset becomes very large. As dimensions increase, data points become increasingly sparse, distances become less meaningful, and many machine learning algorithms require much more data to perform well.

How does it affect KNN?

The Curse of Dimensionality is particularly important for K-Nearest Neighbors (KNN) because KNN relies heavily on distance calculations.

For example, suppose we want to classify customers based on:

- Age

- Income

- Spending score

Finding nearby customers is relatively easy in 3 dimensions.

But if we use 100 or 1,000 features, the data points become spread out across a very large feature space. As a result, even the "nearest" points may actually be quite far away.

Effects on KNN

1. Distances become less meaningful

In high-dimensional spaces, the difference between the nearest and farthest neighbors can become relatively small. Therefore, KNN has difficulty identifying truly similar observations.

2. Data becomes sparse

More dimensions require exponentially more data to maintain the same density of observations.

3. KNN accuracy can decrease

Since KNN depends on finding meaningful neighbors, poor distance relationships can lead to incorrect predictions.

4. Computational cost increases

KNN needs to calculate distances between observations. More features mean more calculations, making prediction slower.


5. Risk of irrelevant features increases

Features that do not contribute to the target can distort distance calculations and make useful neighbors appear farther away.

Example

Imagine classifying customers using only two features:

Age    -   Income

25     -   30,000

27     -   32,000

26     -   31,000

These customers are clearly close to one another.

Now imagine adding 500 additional features, many of which are irrelevant. The distance between customers may now be dominated by these additional features rather than Age and Income.

Consequently, KNN may select poor neighbors, resulting in less accurate predictions.

How to reduce the Curse of Dimensionality

Common approaches include:

- Feature selection – remove irrelevant or redundant features.
- Dimensionality reduction – techniques such as PCA can reduce the number of dimensions.

- Feature scaling – standardize features so that variables with large numerical ranges do not dominate distance calculations.

- Choosing an appropriate K – cross-validation can help select a suitable value of K.

- Removing highly correlated features – reduces unnecessary dimensions.

Therefore, feature selection, dimensionality reduction, and proper scaling are especially important when applying KNN to high-dimensional datasets.


# 3. What is Principal Component Analysis (PCA)? How is it different from feature selection?

- Principal Component Analysis (PCA)

Principal Component Analysis (PCA) is a dimensionality reduction technique used to reduce the number of features in a dataset while retaining as much of the important information (variance) as possible.

PCA transforms the original features into a new set of variables called principal components. These components are combinations of the original features.

How PCA works
1. Standardize the data so that features with different scales are comparable.

2. Calculate the covariance matrix of the features.

3. Find the eigenvectors and eigenvalues of the covariance matrix.

4. Rank the principal components according to the amount of variance they explain.

5. Select the required number of components.

6. Transform the original data into the new lower-dimensional space.

PCA vs. Feature Selection

The main difference is what happens to the original features.

- Feature Selection

Feature selection chooses a subset of the original features and removes the rest.

For example:

Original features:

Age

Income

Education

Height

Weight

Experience

Feature selection might keep:

Age

Income

Experience

The selected features remain unchanged and interpretable.

- PCA

PCA creates new features (principal components) by combining the original features.

For example:

Age + Income + Experience → Principal Component 1

Education + Height + Weight → Principal Component 2

The new components are generally less directly interpretable than the original variables.

Key Differences

| Feature           | PCA                                 | Feature Selection                 |
| ----------------- | ----------------------------------- | --------------------------------- |
| Approach          | Creates new features                | Selects existing features         |
| Original features | Transformed                         | Kept unchanged                    |
| Dimensionality    | Reduced                             | Reduced                           |
| Interpretability  | Usually lower                       | Usually higher                    |
| Information       | Attempts to retain maximum variance | Depends on selection method       |
| Example           | 10 features → 3 components          | 10 features → 3 original features |

Simple Example

Suppose a dataset has 100 features.

a. Feature selection: selects the 20 most useful original features.

b. PCA: transforms all 100 features into, for example, 20 principal components.


# 4.  What are eigenvalues and eigenvectors in PCA, and why are they important?

- Eigenvalues and Eigenvectors in PCA

In Principal Component Analysis (PCA), eigenvalues and eigenvectors are mathematical concepts used to determine the principal components and how much information each component contains.

1. Eigenvectors

An eigenvector represents a direction in the feature space.

In PCA, each eigenvector represents the direction of a principal component—a new axis along which the data varies.

For example, if we have features such as height and weight, PCA finds directions through the data where the observations show the greatest variation.

2. Eigenvalues

An eigenvalue tells us how much variance (information) is captured along its corresponding eigenvector.

Large eigenvalue → captures more variance → more important component.
Small eigenvalue → captures less variance → less important component.

How they work together in PCA

Suppose PCA produces:

| Principal Component | Eigenvalue | Interpretation                    |
| ------------------- | ---------: | --------------------------------- |
| PC1                 |        5.2 | Captures the most variance        |
| PC2                 |        2.1 | Captures the second-most variance |
| PC3                 |        0.5 | Captures less variance            |
| PC4                 |        0.2 | Captures very little variance     |

PCA would generally prioritize PC1 and PC2 because they explain most of the variation in the data.

Why are they important?

Eigenvalues and eigenvectors help PCA:

1. Find the principal directions → eigenvectors provide the directions.

2. Measure the importance of each direction → eigenvalues measure the variance.

3. Reduce dimensionality → components with small eigenvalues can often be discarded.

4. Retain important information → components with large eigenvalues are retained.

Simple way to remember

- Eigenvector = Direction of the principal component

- Eigenvalue = Amount of variance captured by that component

So, if you have 10 original features, PCA might identify 10 principal components. You can then keep only the components with the largest eigenvalues—for example, the first 3—which reduces the dataset from 10 dimensions to 3 dimensions while retaining much of its variation.


# 5.  How do KNN and PCA complement each other when applied in a single pipeline?

- How KNN and PCA Complement Each Other

KNN and PCA can work well together because PCA reduces the number of features before KNN performs its distance-based predictions.

KNN can suffer from the Curse of Dimensionality when a dataset has many features. PCA helps address this by transforming the original high-dimensional data into a smaller number of important principal components.

Typical Pipeline

Original Dataset

       ↓

Data Preprocessing

       ↓

Feature Scaling

       ↓

PCA

       ↓

Reduced-Dimensional Dataset

       ↓

KNN

       ↓

Prediction


- Step-by-step

1. Preprocess the data

Handle missing values and categorical variables as appropriate.

2. Scale the features

Scaling is important because PCA and KNN are both affected by feature magnitude.

For example:

Age:       18–70

Income:    20,000–500,000

Without scaling, Income could have a much larger influence on distance calculations.

3. Apply PCA

PCA transforms the original features into a smaller number of principal components.

For example:

100 original features

        ↓

       PCA

        ↓

10 principal components

The components are selected to retain a desired amount of variance.

4. Apply KNN

KNN uses the reduced set of principal components to calculate distances and identify the nearest neighbors.

For classification:

Nearest neighbors → majority vote → predicted class

For regression:

Nearest neighbors → average target value → predicted value
Why this combination is useful

Why this combination is useful

| PCA                                | KNN                                                       |
| ---------------------------------- | --------------------------------------------------------- |
| Reduces dimensionality             | Makes predictions using nearby observations               |
| Removes some redundant information | Relies on meaningful distances                            |
| Can reduce computational cost      | Distance calculations can be expensive with many features |
| Helps with high-dimensional data   | Can suffer from the Curse of Dimensionality               |

Example

Suppose you have a dataset containing 50 features for predicting whether a customer will default on a loan.

Using KNN directly:

50 features → Distance calculation → KNN → Prediction

There may be too many dimensions, making distances less informative.

With PCA:

50 features

     ↓

 Scaling

     ↓

 PCA

     ↓

 10 components

     ↓

 KNN

     ↓

 Prediction

 KNN now operates in a 10-dimensional space instead of a 50-dimensional space, which can make the distance calculations more manageable.

Important point

PCA does not always improve KNN accuracy. PCA maximizes variance, but the directions with the greatest variance are not necessarily the directions that are most useful for predicting the target. Therefore, the number of principal components should generally be selected using validation/cross-validation, and the PCA transformation must be fitted only on the training data to avoid data leakage.

In short

PCA reduces the dimensionality and complexity of the data, while KNN uses the resulting features to make predictions based on distance.

Therefore, a common combination is:

Scale → PCA → KNN → Evaluate

This can be particularly useful when the original dataset contains many correlated or redundant features.


# 6. Train a KNN Classifier on the Wine dataset with and without feature scaling. Compare model accuracy in both cases. Use the Wine Dataset from sklearn.datasets.load_wine().

- KNN Classifier on the Wine Dataset — With and Without Scaling

The Wine dataset contains 178 samples, 13 numerical features, and 3 classes. Since KNN is distance-based, feature scaling can significantly affect which observations are considered neighbors.

Here is a complete Python program that trains KNN both without scaling and with StandardScaler, then compares their accuracy.

```python
# KNN Classifier on Wine Dataset
# Compare performance with and without feature scaling

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# 1. Load the Wine dataset
wine = load_wine()

X = wine.data
y = wine.target

# 2. Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# ---------------------------------------------------
# MODEL 1: KNN WITHOUT FEATURE SCALING
# ---------------------------------------------------

knn_unscaled = KNeighborsClassifier(n_neighbors=5)

# Train the model
knn_unscaled.fit(X_train, y_train)

# Make predictions
y_pred_unscaled = knn_unscaled.predict(X_test)

# Calculate accuracy
accuracy_unscaled = accuracy_score(y_test, y_pred_unscaled)


# ---------------------------------------------------
# MODEL 2: KNN WITH FEATURE SCALING
# ---------------------------------------------------

# Create the scaler
scaler = StandardScaler()

# Fit scaler ONLY on training data
X_train_scaled = scaler.fit_transform(X_train)

# Use the same transformation on test data
X_test_scaled = scaler.transform(X_test)

# Create KNN classifier
knn_scaled = KNeighborsClassifier(n_neighbors=5)

# Train the model using scaled data
knn_scaled.fit(X_train_scaled, y_train)

# Make predictions
y_pred_scaled = knn_scaled.predict(X_test_scaled)

# Calculate accuracy
accuracy_scaled = accuracy_score(y_test, y_pred_scaled)


# ---------------------------------------------------
# DISPLAY RESULTS
# ---------------------------------------------------

print("KNN Accuracy Comparison")
print("-----------------------")

print(f"Accuracy without scaling : {accuracy_unscaled:.4f}")
print(f"Accuracy with scaling    : {accuracy_scaled:.4f}")

print("\nAccuracy in percentage:")
print(f"Without scaling : {accuracy_unscaled * 100:.2f}%")
print(f"With scaling    : {accuracy_scaled * 100:.2f}%")
```

The exact values can vary if you change the train/test split, K, or other settings.

Comparison

| Model | Feature Scaling | Accuracy |
| ----- | --------------- | -------: |
| KNN   | No              |  ~72.22% |
| KNN   | Yes             |  ~94.44% |

Why does scaling improve KNN?

KNN calculates distances between observations. The Wine dataset contains features measured on different scales. For example, some features have much larger numerical ranges than others. Without scaling, large-range features can dominate the distance calculation.

StandardScaler transforms each feature approximately so that it has:

- Mean = 0
- Standard deviation = 1

This allows the features to contribute more comparably to the distance calculation.

Conclusion: In this experiment, feature scaling substantially improves KNN performance because KNN depends directly on distances. Therefore, scaling is generally important when using KNN with features that have different scales.


# 7. Train a PCA model on the Wine dataset and print the explained variance ratio of each principal component.

-  PCA on the Wine Dataset

The Wine dataset contains 178 samples and 13 numerical features. Since the features have different scales, it is good practice to standardize them before applying PCA.

The following program fits PCA with all 13 components and prints the explained variance ratio of each principal component.

```python
# PCA on Wine Dataset
# Print explained variance ratio of each principal component

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Load the Wine dataset
wine = load_wine()

X = wine.data

# 2. Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Apply PCA
# Keep all 13 principal components
pca = PCA(n_components=13)
X_pca = pca.fit_transform(X_scaled)

# 4. Print explained variance ratio
print("Explained Variance Ratio of Each Principal Component:")
print("----------------------------------------------------")

for i, variance in enumerate(pca.explained_variance_ratio_, start=1):
    print(f"PC{i}: {variance:.4f} ({variance * 100:.2f}%)")

# 5. Print total explained variance
print("\nTotal Explained Variance:",
      pca.explained_variance_ratio_.sum())
```

Interpretation

- PC1 explains about 36.16% of the variance.

- PC2 explains about 19.21%.

- Together, PC1 and PC2 explain approximately 55.37%.

- The first few components therefore capture a substantial portion of the
variation in the original 13 features.

In short: PCA converts the original 13 features into 13 principal components and ranks them according to the amount of variance they explain. Components with larger explained variance ratios contain more of the dataset's overall variation.


# 8. Train a KNN Classifier on the PCA-transformed dataset (retain top 2 components). Compare the accuracy with the original dataset.

- KNN with PCA on the Wine Dataset

We can compare:

1. KNN on the original 13 features
2. KNN on only the top 2 PCA components

Because KNN is distance-based, we should standardize the features before PCA and KNN. The Wine features have different scales, and scikit-learn specifically demonstrates the importance of scaling for KNN on this dataset.

```python
# KNN with and without PCA on the Wine Dataset

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1. Load Wine dataset
wine = load_wine()

X = wine.data
y = wine.target

# 2. Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# ---------------------------------------------------
# STEP 1: Standardize the data
# ---------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ---------------------------------------------------
# MODEL 1: KNN on original 13 features
# ---------------------------------------------------

knn_original = KNeighborsClassifier(n_neighbors=5)

knn_original.fit(X_train_scaled, y_train)

y_pred_original = knn_original.predict(X_test_scaled)

accuracy_original = accuracy_score(
    y_test, y_pred_original
)

# ---------------------------------------------------
# STEP 2: Apply PCA and retain 2 components
# ---------------------------------------------------

pca = PCA(n_components=2)

X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

# ---------------------------------------------------
# MODEL 2: KNN on 2 PCA components
# ---------------------------------------------------

knn_pca = KNeighborsClassifier(n_neighbors=5)

knn_pca.fit(X_train_pca, y_train)

y_pred_pca = knn_pca.predict(X_test_pca)

accuracy_pca = accuracy_score(
    y_test, y_pred_pca
)

# ---------------------------------------------------
# RESULTS
# ---------------------------------------------------

print("KNN Accuracy Comparison")
print("-----------------------")

print(f"Original 13 features : {accuracy_original:.4f}")
print(f"Top 2 PCA components : {accuracy_pca:.4f}")

print("\nAccuracy in percentage:")
print(f"Original dataset : {accuracy_original * 100:.2f}%")
print(f"PCA dataset      : {accuracy_pca * 100:.2f}%")

# Explained variance of the two components
print("\nExplained variance:")
print(f"PC1: {pca.explained_variance_ratio_[0]:.4f}")
print(f"PC2: {pca.explained_variance_ratio_[1]:.4f}")

print(
    f"Total: {pca.explained_variance_ratio_.sum():.4f}"
)
```

The exact result can vary with the split or other parameter choices.

Comparison

The two PCA components retain about 55.37% of the total variance, so some information contained in the remaining 11 components is discarded.

Conclusion

In this experiment, the original 13-feature dataset gives slightly higher accuracy than using only the top 2 PCA components. However, PCA reduces the dimensionality from 13 features to just 2, which can make the data much easier to visualize and can reduce computational complexity.

This illustrates an important point: PCA does not necessarily improve classification accuracy. It trades some information for dimensionality reduction. Since KNN makes predictions using the nearest neighbors and distance calculations, reducing dimensions can be useful, but keeping only two components may discard information that helps classification.


# 9. Train a KNN Classifier with different distance metrics (euclidean, manhattan) on the scaled Wine dataset and compare the results.

- KNN with Different Distance Metrics on the Wine Dataset

KNN can use different distance metrics to determine which observations are nearest. In scikit-learn, Euclidean distance corresponds to Minkowski distance with \(p=2\), while Manhattan distance corresponds to \(p=1\).

Since the Wine features are on different scales, we first apply StandardScaler, fitting it only on the training data.

```python
# KNN with Euclidean and Manhattan Distance
# Wine Dataset

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1. Load Wine dataset
wine = load_wine()

X = wine.data
y = wine.target

# 2. Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# 3. Scale the features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ------------------------------------------------
# KNN with Euclidean Distance
# ------------------------------------------------

knn_euclidean = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean"
)

knn_euclidean.fit(X_train_scaled, y_train)

y_pred_euclidean = knn_euclidean.predict(X_test_scaled)

accuracy_euclidean = accuracy_score(
    y_test, y_pred_euclidean
)

# ------------------------------------------------
# KNN with Manhattan Distance
# ------------------------------------------------

knn_manhattan = KNeighborsClassifier(
    n_neighbors=5,
    metric="manhattan"
)

knn_manhattan.fit(X_train_scaled, y_train)

y_pred_manhattan = knn_manhattan.predict(X_test_scaled)

accuracy_manhattan = accuracy_score(
    y_test, y_pred_manhattan
)

# ------------------------------------------------
# Compare Results
# ------------------------------------------------

print("KNN Distance Metric Comparison")
print("--------------------------------")

print(f"Euclidean Accuracy : {accuracy_euclidean:.4f}")
print(f"Manhattan Accuracy : {accuracy_manhattan:.4f}")

print("\nAccuracy in percentage:")
print(f"Euclidean : {accuracy_euclidean * 100:.2f}%")
print(f"Manhattan : {accuracy_manhattan * 100:.2f}%")
```

Explanation

Euclidean distance measures the straight-line distance between two observations:

$$ d=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2} $$

Manhattan distance measures the sum of absolute differences:

$$ d=\sum_{i=1}^{n}|x_i-y_i| $$

Because the data has been standardized, features contribute on comparable scales. The choice of metric can still change which observations are considered nearest, so it is useful to test both when developing a KNN model. Scikit-learn allows the metric to be specified directly through the metric parameter.

Conclusion

In this particular experiment, both distance metrics can produce very similar accuracy. This does not mean the metrics are mathematically equivalent; rather, for this Wine dataset and this particular train/test split and \(K=5\), they lead to similar classification results.

For a more reliable comparison, you could evaluate both metrics using 5-fold cross-validation rather than relying on a single train/test split.

-  PCA + KNN Pipeline for Cancer Classification

A PCA + KNN pipeline can be useful for high-dimensional gene-expression data because gene-expression datasets often have thousands of features but relatively few patient samples. The main goal is to reduce noise and dimensionality while preserving useful patterns before applying KNN.

1. Use PCA to reduce dimensionality

First, preprocess the gene-expression data:

1. Handle missing values if present.

2. Remove features with extremely low variance.

3. Standardize the gene-expression features so that they are on a comparable scale.

4. Fit PCA only on the training data to avoid data leakage.

PCA transforms the original features into a smaller set of principal components (PCs) that capture directions of variation in the data.

For example:

20,000 gene features

        ↓

   Standardization

        ↓

        PCA

        ↓

  50–100 components

        ↓

       KNN

        ↓

 Cancer classification

This reduces the dimensionality dramatically and can make KNN's distance calculations more meaningful.

2. Decide how many components to keep

I would not simply choose an arbitrary number of components.

Several approaches can be considered:

A. Explained variance

Examine the cumulative explained variance and select enough components to retain, for example, 90–95% of the variance.

PC1 + PC2 + ... + PCk


          ↓
90–95% variance retained

However, explained variance alone should not determine the final choice because the directions with the most variance are not necessarily the most useful for predicting cancer type.

B. Cross-validation

A better approach is to test different numbers of components, such as:

10, 20, 50, 100, 200

and select the number that performs well under stratified cross-validation.

The number of PCA components should therefore be treated as a hyperparameter.

3. Use KNN after dimensionality reduction

Once PCA has reduced the feature space, train a KNN classifier on the principal components.

For example:

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=50)),
    ("knn", KNeighborsClassifier(n_neighbors=5))
])
```

KNN classifies a new patient by finding the K most similar patients in the reduced PCA space.

For example, with K = 5:

3 neighbors → Cancer Type A

2 neighbors → Cancer Type B

Prediction → Cancer Type A

I would tune both:

- Number of PCA components

- n_neighbors

- Distance metric, such as Euclidean or Manhattan

- Potentially distance weighting (weights="distance")

using cross-validation.

4. Evaluate the model properly

Accuracy alone is not sufficient for a biomedical classification problem, particularly if cancer classes are imbalanced.

I would use:

Primary metrics

- Balanced accuracy

- Macro F1-score

- Precision

- Recall/sensitivity

- Specificity

- Confusion matrix

For multiple cancer types, I would report metrics for each cancer class, rather than only one overall accuracy.

I would also use stratified cross-validation and keep a completely independent test set for the final evaluation.

A particularly important consideration for gene-expression data is data leakage. PCA must be fitted only on the training portion of each cross-validation fold—not on the entire dataset before cross-validation.

A pipeline such as:

```python
Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA()),
    ("knn", KNeighborsClassifier())
])
```

helps ensure that preprocessing is performed correctly within each training fold.

| Biomedical reliability          | Independent test set and robust validation           |


Important limitation

I would not claim that PCA + KNN is automatically a clinically robust or production-ready solution simply because it performs well on one dataset. Gene-expression datasets can contain strong batch effects, population differences, measurement variation, and dataset-specific patterns.

Before real-world clinical use, the model should ideally be validated on an independent external cohort and assessed for reproducibility and clinically relevant errors.

Stakeholder summary

The PCA + KNN pipeline provides a computationally efficient approach for high-dimensional gene-expression classification. PCA reduces thousands of correlated gene features into a manageable number of components, while KNN performs classification based on similarity between patients in this reduced space. The number of components and KNN parameters would be selected using cross-validation, and performance would be evaluated using class-sensitive metrics and an independent test set. Most importantly, preprocessing would occur inside the validation pipeline to prevent data leakage. External validation would then be required before considering clinical deployment.

This makes the approach simple, interpretable at the pipeline level, computationally manageable, and appropriately cautious for biomedical data, while recognizing that strong validation is essential before drawing clinical conclusions.
